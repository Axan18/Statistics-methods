```kotlin
import kotlin.random.Random
import kotlin.math.ln

fun expTime(lambda: Double): Double =
  -ln(1 - Random.nextDouble()) / lambda

data class State(val time: Double, val queue: Int, val done: Int)

fun simulate(lambdaA: Double, lambdaS: Double, Tmax: Double): List<State> {
  var time = 0.0
  var queue = 0
  var done = 0
  var serverBusy = false

  var nextArrival = expTime(lambdaA)
  var nextServiceEnd = Double.POSITIVE_INFINITY

  val history = mutableListOf<State>()

  while (time < Tmax) {
    if (nextArrival < nextServiceEnd) {
      time = nextArrival
      if (!serverBusy) {
        serverBusy = true
        nextServiceEnd = time + expTime(lambdaS)
      } else {
        queue++
      }
      nextArrival += expTime(lambdaA)
    } else {
      time = nextServiceEnd
      done++
      if (queue > 0) {
        queue--
        nextServiceEnd = time + expTime(lambdaS)
      } else {
        serverBusy = false
        nextServiceEnd = Double.POSITIVE_INFINITY
      }
    }
    history.add(State(time, queue, done))
  }
  return history
}
```

```kotlin
val lambdaA = 0.05
val lambdaS = 1.0/15
val traj = simulate(lambdaA, lambdaS, 500.0)

val df = dataFrameOf(
  "time" to traj.map { it.time },
  "queue" to traj.map { it.queue },
  "done" to traj.map { it.done }
)

plot(df) {
  line {
    x("time")
    y("queue")
  }
  layout { title = "Liczba zadań w kolejce" }
}

```

```kotlin
plot(df) {
  line {
    x("time")
    y("done")
  }
  layout { title = "Liczba wykonanych zadań" }
}
```
