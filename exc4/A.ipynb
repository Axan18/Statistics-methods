```kotlin
import kotlin.random.Random
import kotlin.math.ln

val nTrajectories = 10_000
val lambda = 1.0
public fun nextInterArrivalTime(lambda: Double): Double {
  val u = Random.nextDouble()
  return -ln(1.0 - u) / lambda
}
```

```kotlin
val times = mutableListOf(0.0)
val states = mutableListOf(0)
var currentTime = 0.0
while (currentTime < 20.0) {
  currentTime += nextInterArrivalTime(lambda)
  times.add(currentTime)
  states.add(states.last() + 1)
}

plot {
  step {
    x(times)
    y(states)
  }
  layout { title = "Przykładowa trajektoria procesu Poissona" }
}
```

```kotlin
val checkPoints = listOf(1.0, 20.0, 90.0)
val results = mutableListOf<Map<String, Any>>()

repeat(nTrajectories) {
  var t = 0.0
  var count = 0
  val trajectoryCounts = mutableMapOf<Double, Int>()

  for (cp in checkPoints) {
    while (t <= cp) {
      val dt = nextInterArrivalTime(lambda)
      if (t + dt > cp) break
      t += dt
      count++
    }
    trajectoryCounts[cp] = count
  }

  results.add(mapOf(
    "t1" to trajectoryCounts[1.0]!!,
    "t20" to trajectoryCounts[20.0]!!,
    "t90" to trajectoryCounts[90.0]!!
  ))
}
val dfResults = results.toDataFrame()
```

```kotlin
import org.jetbrains.kotlinx.kandy.ir.Plot

fun poissonTheo(k: Int, lambdaT: Double): Double {
  if (k < 0) return 0.0
  var logFactorial = 0.0
  for (i in 1..k) logFactorial += ln(i.toDouble())
  return exp(k * ln(lambdaT) - lambdaT - logFactorial)
}

fun plotDistribution(time: Double): Plot {
  val lambda = 1.0
  val colName = "t${time.toInt()}"
  val nTraj = dfResults.rowsCount()

  val countsMap = dfResults[colName].valueCounts()

  val kCol = countsMap.getColumn(0)
  val nCol = countsMap.getColumn(1)

  val kList = kCol.toList().map { it as Int }
  val nList = nCol.toList().map { (it as Number).toInt() }


  val empiricalDf = dataFrameOf(
    "k" to kList,
    "empiryczny" to nList.map { it.toDouble() / nTraj.toDouble() }
  )

  val theoreticalDf = dataFrameOf(
    "k" to kList,
    "teoretyczny" to kList.map { poissonTheo(it, lambda * time) }
  )

  val combinedDf = empiricalDf.innerJoin(theoreticalDf) { it["k"] }

  return plot(combinedDf) {
    bars {
      x("k")
      y("empiryczny")
      fillColor = Color.BLUE
      alpha = 0.5
    }
    points {
      x("k")
      y("teoretyczny")
      color = Color.RED
      symbol = Symbol.CIRCLE_FILLED
    }
    layout {
      title = "Rozkład w czasie t=$time vs Teoretyczny Poisson(λt)"
    }
  }
}
```

```kotlin
plotDistribution(1.0)
```

```kotlin
plotDistribution(20.0)
```

```kotlin
plotDistribution(90.0)
```
