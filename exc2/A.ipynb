```kotlin
import org.jetbrains.kotlinx.multik.default.linalg.DefaultLinAlgEx.dotMM

val initialMatrix = mk.ndarray(mk[mk[0.64,0.32, 0.04],mk[0.4,0.5, 0.1], mk[0.25, 0.5, 0.25]])

var matrix = initialMatrix
repeat(5000){
  matrix = dotMM(matrix, initialMatrix)
}
println(matrix)
```

```kotlin
val epsilon = 1e-16
var diff: Double? = null
var counter = 0
matrix = initialMatrix
val stateZero = mutableListOf<Double>()
val stateOne = mutableListOf<Double>()
val stateTwo = mutableListOf<Double>()
do{
  counter+=1
  stateZero.add(matrix[0][0])
  stateOne.add(matrix[0][1])
  stateTwo.add(matrix[0][2])
  val prev = matrix.copy()
  matrix = dotMM(matrix, initialMatrix)
  diff = DefaultLinAlg.linAlgEx.norm(matrix - prev, Norm.Fro)
}while (diff!! > epsilon)
println("Iterations: $counter")
println(matrix)
```

```kotlin
import org.jetbrains.kotlinx.kandy.ir.scale.NonPositionalScale
import org.jetbrains.letsPlot.core.spec.plotson.scale
import org.jetbrains.letsPlot.scale.scaleDiscrete

val df = dataFrameOf("Iteration" to stateZero.indices.map{it},
  "P0" to stateZero,
  "P1" to stateOne,
  "P2" to stateTwo
)
df.plot{
  line{
    x("Iteration")
    y("P0"){
      scale= continuous(0.0, 1.0)
    }
    color = Color.hex("#ff0000")
  }
  line{
    x("Iteration")
    y("P1"){
      scale= continuous(0.0, 1.0)
    }
    color = Color.hex("#f0ff00")
  }
  line{
    x("Iteration")
    y("P2"){
      scale= continuous(0.0, 1.0)
    }
    color = Color.hex("#f00fff")
  }
  hLine{
    yIntercept.constant(matrix[0][0])
    color = Color.hex("#00ff00")
    type = LineType.DASHED
  }
  hLine{
    yIntercept.constant(matrix[0][1])
    color = Color.hex("#00ff00")
    type = LineType.DASHED
  }
  hLine{
    yIntercept.constant(matrix[0][2])
    color = Color.hex("#00ff00")
    type = LineType.DASHED
  }
}
```
