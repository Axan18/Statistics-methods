```kotlin
import org.jetbrains.kotlinx.kandy.ir.Plot
import kotlin.random.Random

fun simulation(start: Int): Plot {
  val matrix = mk.ndarray(mk[mk[0.64, 0.32, 0.04], mk[0.4, 0.5, 0.1], mk[0.25, 0.5, 0.25]])

  val stateZero = mutableListOf<Double>()
  val stateOne = mutableListOf<Double>()
  val stateTwo = mutableListOf<Double>()
  var counts = mutableMapOf(0 to 0.0, 1 to 0.0, 2 to 0.0)
  val totalSteps = 10000

  var current = start
  for (step in 1..totalSteps) {
    counts[current] = counts[current]!! + 1

    stateZero.add(counts[0]!! / step)
    stateOne.add(counts[1]!! / step)
    stateTwo.add(counts[2]!! / step)

    val r = Random.nextDouble()
    current = when {
      r < matrix[current, 0] -> 0
      r < matrix[current, 0] + matrix[current, 1] -> 1
      else -> 2
    }
  }

  val df = dataFrameOf(
    "Step" to (1..totalSteps).toList(),
    "0" to stateZero,
    "1" to stateOne,
    "2" to stateTwo
  )
  return df.plot {
    layout {
      title = "Start $start"
    }
    line {
      x("Step")
      y("0")
      color = Color.hex("#ff0000")
    }
    line {
      x("Step")
      y("1")
      color = Color.hex("#00ff00")
    }
    line {
      x("Step")
      y("2")
      color = Color.hex("#0000ff")
    }
  }
}
```

```kotlin
simulation(0)
```

```kotlin
simulation(1)
```

```kotlin
simulation(2)
```
