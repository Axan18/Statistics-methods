```kotlin
import kotlin.math.pow
import kotlin.random.Random
%use dataframe
%use kandy
```

```kotlin
import org.apache.commons.math3.distribution.ConstantRealDistribution
import org.apache.commons.math3.distribution.ExponentialDistribution
import org.apache.commons.math3.distribution.UniformRealDistribution
import org.jetbrains.kotlinx.statistics.distribution.NormalDistribution

val normalDist = mutableListOf<List<Double>>()
val expDist = mutableListOf<List<Double>>()
val uniDist = mutableListOf<List<Double>>()

for (i in 0..<100000) {
  normalDist.add(NormalDistribution().sample(10).toList())
  expDist.add(ExponentialDistribution(1.0).sample(10).toList())
  uniDist.add(UniformRealDistribution().sample(10).toList())
}

val df = dataFrameOf(
  "normal" to normalDist.map { it.sum() },
  "exp" to expDist.map { it.sum() },
  "uniDist" to uniDist.map { it.sum() }
)
```

```kotlin
df.plot{
  statBin("normal"){
    bars{
      x(Stat.x)
      y(Stat.count)
    }
  }
}
```

```kotlin
df.plot{
  statBin("exp"){
    bars{
      x(Stat.x)
      y(Stat.count)
    }
  }
}
```

```kotlin
df.plot{
  statBin("uniDist"){
    bars{
      x(Stat.x)
      y(Stat.count)
    }
  }
}
```

```kotlin
val normal = NormalDistribution().sample(100_000).toList().sorted()
val exp = ExponentialDistribution(1.0).sample(100_000).toList().sorted()
val uni = UniformRealDistribution().sample(100_000).toList().sorted()

val df = dataFrameOf(
  "normal" to normal,
  "exp" to exp,
  "const" to uni
)

fun consecutiveDiffs(sorted: List<Double>): List<Double> =
  sorted.zipWithNext { a, b -> b - a }
```

```kotlin
val diffs = consecutiveDiffs(normal)

plot {
  histogram(diffs)
  layout {
    title = "Histogram różnic (skala liniowa)"
    xAxisLabel = "Różnica między kolejnymi liczbami"
    yAxisLabel = "Liczność"
  }
}
```

```kotlin
import org.jetbrains.kotlinx.kandy.ir.scale.PositionalScale

plot {
  histogram(diffs)
  layout {
    title = "Histogram różnic (skala logarytmiczna)"
    xAxisLabel = "Różnica między kolejnymi liczbami"
    yAxisLabel = "Liczność"
  }
  y{
    scale = continuous(transform = Transformation.LOG10)
  }
}
```
