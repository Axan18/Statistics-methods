```kotlin
import kotlin.random.Random

fun boxMuller() : Pair<Double, Double>{
  val x1 = Random.nextFloat()
  val x2 = Random.nextFloat()
  val y1 = sqrt(-2.0 * ln(x1)) * cos(2.0 * Math.PI * x2)
  val y2 = sqrt(-2.0 * ln(x1)) * sin(2.0 * Math.PI * x2)
  return y1 to y2
}

fun normalPDF(x: Double): Double {
  return (1.0 / sqrt(2 * PI)) * exp(-x * x / 2)
}
```

```kotlin
import org.jetbrains.kotlinx.dataframe.api.dataFrameOf
import org.jetbrains.kotlinx.kandy.ir.Plot

fun main(): Plot {
  val n = 10000
  val samples = List(n) { boxMuller() }.flatMap { (st, nd) -> listOf(st, nd) }

  val minX = -3.0
  val maxX = 3.0
  val bins = 30
  val binWidth = (maxX - minX) / bins

  val counts = IntArray(bins)
  for (v in samples) {
    if (v >= minX && v < maxX) {
      val idx = ((v - minX) / binWidth).toInt().coerceIn(0, bins - 1)
      counts[idx]++
    }
  }

  val centers = (0 until bins).map { i -> minX + (i + 0.5) * binWidth }
  val densities = counts.map { it.toDouble() / (n*2 * binWidth) }

  val histDF = dataFrameOf(
    "x" to centers,
    "density" to densities
  )

  val xs = (0..<1000).map { i -> minX + i * (maxX - minX) / 1000.0 }
  val ys = xs.map { normalPDF(it) }

  val pdfDF = dataFrameOf(
    "x" to xs,
    "pdf" to ys
  )

  val plt = plot {
    bars {
      x(histDF["x"])
      y(histDF["density"])
      alpha = 0.6
      fillColor = Color.hex("#00ffff")
    }

    line {
      x(pdfDF["x"])
      y(pdfDF["pdf"])
      color = Color.hex("#f0f0f0")
      width = 2.0
    }
  }
  return plt
}
```

```kotlin
main()
```
